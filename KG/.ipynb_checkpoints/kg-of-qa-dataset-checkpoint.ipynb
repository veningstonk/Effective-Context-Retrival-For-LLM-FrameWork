```python
import json
import pandas as pd
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
file_path = "/kaggle/input/squad-v20/train-v2.0.json"
with open(file_path, "r", encoding="utf-8") as f:
    squad_data = json.load(f)["data"]
```

```python
rdf_triples = []  # (Question, answered_by, Answer)
ner_entities = []  # (Entity, Label)
```

```python
max_articles = 2
```

```python
for article in squad_data[:max_articles]:
    for paragraph in article["paragraphs"]:
        context = paragraph["context"]
        doc = nlp(context)

        # Extract Named Entities from the context
        for ent in doc.ents:
            ner_entities.append((ent.text, ent.label_))

        # Extract QA pairs
        for qa in paragraph["qas"]:
            question = qa["question"]
            if not qa["is_impossible"]:
                for ans in qa["answers"]:
                    answer = ans["text"]
                    rdf_triples.append((question, "answered_by", answer))
            else:
                rdf_triples.append((question, "answered_by", "No Answer"))
```

```python
triples_df = pd.DataFrame(rdf_triples, columns=["Subject", "Predicate", "Object"])
entities_df = pd.DataFrame(ner_entities, columns=["Entity", "Label"]).drop_duplicates()
```

```python
triples_df.to_csv("/kaggle/working/neo4j_rdf_triples.csv", index=False)
entities_df.to_csv("/kaggle/working/ner_entities.csv", index=False)
```

```python
print("Sample RDF Triples (for Neo4j or KG tools):")
display(triples_df.head())

print("\nSample Named Entities Extracted from Context:")
display(entities_df.head())
```

```python



import json
import pandas as pd
import spacy
import networkx as nx
import matplotlib.pyplot as plt


nlp = spacy.load("en_core_web_sm")


file_path = "/kaggle/input/squad-v20/train-v2.0.json"
with open(file_path, "r", encoding="utf-8") as f:
    squad_data = json.load(f)["data"]


rdf_triples = []      
ner_entities = []     
sample_qas = []       


max_articles = 2


for article in squad_data[:max_articles]:
    for paragraph in article["paragraphs"]:
        context = paragraph["context"]
        doc = nlp(context)

        # Named Entity Recognition from context
        for ent in doc.ents:
            ner_entities.append((ent.text, ent.label_))

        for qa in paragraph["qas"]:
            question = qa["question"]
            if not qa["is_impossible"]:
                for ans in qa["answers"]:
                    answer = ans["text"]
                    rdf_triples.append((question, "answered_by", answer))
                    if len(sample_qas) < 20:
                        sample_qas.append((question, answer))
            else:
                rdf_triples.append((question, "answered_by", "No Answer"))
                if len(sample_qas) < 20:
                    sample_qas.append((question, "No Answer"))


triples_df = pd.DataFrame(rdf_triples, columns=["Subject", "Predicate", "Object"])
entities_df = pd.DataFrame(ner_entities, columns=["Entity", "Label"]).drop_duplicates()


triples_df.to_csv("/kaggle/working/neo4j_rdf_triples.csv", index=False)
entities_df.to_csv("/kaggle/working/ner_entities.csv", index=False)


print(" Sample RDF Triples (for Neo4j or KG tools):")
display(triples_df.head())

print("\n Sample Named Entities Extracted from Context:")
display(entities_df.head())


G = nx.DiGraph()

for subj, pred, obj in rdf_triples:
    G.add_node(subj, type="Question")
    G.add_node(obj, type="Answer")
    G.add_edge(subj, obj, label=pred)


subgraph_nodes = set()
for q, a in sample_qas:
    subgraph_nodes.add(q)
    subgraph_nodes.add(a)

SG = G.subgraph(subgraph_nodes)

plt.figure(figsize=(18, 12))
pos = nx.spring_layout(SG, k=0.5, seed=42)

color_map = []
for node in SG.nodes():
    node_type = G.nodes[node]["type"]
    if node_type == "Question":
        color_map.append("lightblue")
    elif node_type == "Answer":
        color_map.append("salmon")
    else:
        color_map.append("lightgray")

nx.draw(SG, pos, with_labels=True, node_color=color_map, node_size=2800,
        font_size=8, font_weight="bold", edge_color="gray")

edge_labels = nx.get_edge_attributes(SG, 'label')
nx.draw_networkx_edge_labels(SG, pos, edge_labels=edge_labels, font_color="red", font_size=8)

plt.title(" Knowledge Graph Sample from RDF Triples (20 QA Pairs)", fontsize=14)
plt.axis("off")


import warnings
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    plt.tight_layout()

plt.show()


print("20 Sample QA Pairs Extracted from RDF Triples:\n")
for i, (q, a) in enumerate(sample_qas, 1):
    print(f"* Q{i}: {q}")
    print(f"   A{i}: {a}\n")
```
